```python
print("Avengers Endgame - What If Episode ")

event = input(
    "Enter event from the list "
    "(meet, captain_not_meet_peggy, ironman_not_get_stones, thor_snaps): "
)

if event == "meet":
    print("The Avengers work together, collect the Infinity Stones, and defeat Thanos.")

elif event == "captain_not_meet_peggy":
    print("Captain America returns to the present and continues as an Avenger.")

elif event == "ironman_not_get_stones":
    print("Thanos wins again because Iron Man cannot perform the final snap.")

elif event == "thor_snaps":
    print("Thor performs the final snap, and the Avengers win without losing Iron Man.")

else:
    print("Invalid event. Please choose a valid option.")
```
