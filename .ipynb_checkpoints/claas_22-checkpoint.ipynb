```python
mylist = [1,2,3,4,5,[1,2,3]]
count=0
for i in mylist:
    print("idx" , count ,"value", i)
    count += 1    
```

```python
mylist = [[1,2,3,4,5],[1,2,3]]
for x in mylist:
    for y in x:
        print(y)
```
