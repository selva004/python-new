```python
ticket = int(input("Select the rate needed to see the movie (max 250): "))

if ticket <= 100:
    print("Welcome to Kollywood movie")
elif ticket > 100 and ticket < 150:
    print("Welcome to Hollywood movie")
elif ticket > 200 and ticket <= 250:
    print("Welcome to 3D movie")
else:
    print("Invalid ticket rate")
```
